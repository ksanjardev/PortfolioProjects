```python
#  Import libraries

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
plt.style.use('ggplot')
from matplotlib.pyplot import figure

%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12, 8)

df = pd.read_csv(r'D:\Data analytics\Pandas\movies.csv')
pd.set_option('display.max.columns', 8000)
df
```

```python
#  Let's see if there is any nulls

for col in df.columns:
    percent_missing= np.mean(df[col].isnull())
    print('{} - {:.2f}%'.format(col, percent_missing))
```

```python
df['budget'] = df['budget'].fillna(0).astype('int64')
df['gross'] = df['gross'].fillna(0).astype('int64')
```

```python
df
```

```python
df.info()
```

```python
# Create correct Year Column
df['year_correct'] = df['released'].str.extract(r'(\d{4})')
```

```python
df
```

```python
df = df.sort_values(by=['gross'], inplace = False, ascending=False)
```

```python
# Drop any duplicates

df = df.drop_duplicates()
df
```

```python
# Budget high correlation
# Company high correlation
# Scatter plot with budget vs gross

plt.scatter(x=df['budget'], y=df['gross'])
plt.title('Budget vs Gross Earnings')
plt.xlabel('Gross Earnings')
plt.ylabel('Budget for Film')
plt.show()
```

```python
df.head()
```

```python
# Plot budget vs gross using seaborn
sns.regplot(x='budget', y='gross', data = df, scatter_kws = {"color":"red"}, line_kws = {"color":"blue"})
```

```python
df.corr(numeric_only=True, method = 'pearson')  # pearson, kendall, spearman
```

```python
correlation_matrix = df.corr(numeric_only=True)
plt.title('Correlation Matrix For Numeric Features')
plt.xlabel('Movie Features')
plt.ylabel('Movie Features')
sns.heatmap(correlation_matrix, annot=True)
```

```python
# Numerizinig companies
df_numerized = df

for col_name in df_numerized.columns:
    if(df_numerized[col_name].dtype == 'str'):
        df_numerized[col_name] = df_numerized[col_name].astype('category')
        df_numerized[col_name] = df_numerized[col_name].cat.codes
df_numerized
```

```python
correlation_matrix = df_numerized.corr(numeric_only=True)
plt.title('Correlation Matrix For Numeric Features')
plt.xlabel('Movie Features')
plt.ylabel('Movie Features')
sns.heatmap(correlation_matrix, annot=True)
```

```python
correlation_mat = df_numerized.corr(numeric_only=True)
corr_pairs = correlation_mat.unstack()
corr_pairs
```

```python
sorted_pairs = corr_pairs.sort_values()
sorted_pairs
```

```python
high_corr = sorted_pairs[(sorted_pairs) > 0.5]
high_corr
```
